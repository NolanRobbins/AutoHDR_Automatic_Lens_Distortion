# Exploratory Data Analysis
Let's load a few distorted and corrected image pairs from our dataset to see what we are working with.

```python
import os
import glob
import cv2
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

# Find all original (distorted) and generated (corrected) images
train_dir = '../data/train'
orig_images = sorted(glob.glob(os.path.join(train_dir, '*_original.jpg')))
gen_images = sorted(glob.glob(os.path.join(train_dir, '*_generated.jpg')))

print(f"Found {len(orig_images)} original images and {len(gen_images)} corrected images.")
```

```python
def show_pair(idx):
    orig_path = orig_images[idx]
    # Assuming the names match except for the suffix
    gen_path = orig_path.replace('_original.jpg', '_generated.jpg')
    
    if not os.path.exists(gen_path):
        print(f"Corrected image not found for {orig_path}")
        return
        
    orig_img = cv2.imread(orig_path)
    orig_img = cv2.cvtColor(orig_img, cv2.COLOR_BGR2RGB)
    
    gen_img = cv2.imread(gen_path)
    gen_img = cv2.cvtColor(gen_img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(15, 7))
    plt.subplot(1, 2, 1)
    plt.title('Distorted (Original)')
    plt.imshow(orig_img)
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.title('Corrected (Generated)')
    plt.imshow(gen_img)
    plt.axis('off')
    plt.show()

# Show the first pair
show_pair(0)
```
